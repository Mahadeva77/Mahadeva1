```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("students.csv")
print("First 5 records:")
print(df.head())
```

```python
print("Last 3 records:")
print(df.tail(3))
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["Marks"],bins=10)
plt.title("Distribution Of Student's Marks")
plt.xlabel("Marks----->")
plt.ylabel("Number of students----->")
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.boxplot(x=df["StudyHours"])
plt.title("Box Plot of Student marks")
plt.xlabel("StudyHours")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x="StudyHours",y="Marks",data=df,hue="Result",palette={"Pass":"green","Fail":"red"},s=70)
sns.regplot(x="StudyHours",y="Marks",data=df,scatter=False)
plt.title("Study Hours vs Marks by Result")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.legend(title="Result")
plt.show()
```

```python
sns.pairplot(df[["StudyHours","Attendance","PrevMarks","Marks"]])
plt.show()
```

```python
correlation=df[["StudyHours","Attendance","PrevMarks","Marks"]].corr()
print("\nCorrelation Matrix:\n")
print(correlation)
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(correlation,annot=True)
plt.title("Correlation Heatmap")
plt.show()
```
